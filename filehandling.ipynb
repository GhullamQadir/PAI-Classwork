```python
def Addrecord():
    name = input('Enter your name: ')
    age = input('Enter your age: ')
    rollno = input('Enter your roll number: ')
    gpa = input('Enter your GPA: ')
    file = open('file.txt','a')
    file.write(f'{name},{age},{rollno},{gpa}\n')
    file.close()
```

```python
def showAllrecords():
    with open('file.txt','r') as file:
        for recordno, line in enumerate(file, start=1):
            record = line.strip().split(',')
            print(f'****Record No: {recordno}*****')
            print(f'Name: {record[0]} \n Age: {record[1]} \n Roll No: {record[2]} \n GPA: {record[3]}')
```

```python
def showrecord(rollno):
    with open('file.txt','r') as file:
        for recordno, line in enumerate(file, start=1):
            record = line.strip().split(',')
            if record[2] == rollno:
                print(f'****Record No: {recordno}*****')
                print(f'Name: {record[0]} \n Age: {record[1]} \n Roll No: {record[2]} \n GPA: {record[3]}')
```

```python
while True:
    print('1. Add Record')
    print('2. Show All Records')
    print('3. Show Record by Roll Number')
    print('4. Exit')
    choice = input('Enter your choice: ')
    if choice == '1':
        Addrecord()
    elif choice == '2':
        showAllrecords()
    elif choice == '3':
        rollno = input('Enter roll number: ')
        showrecord(rollno)
    elif choice == '4':
        break
    else:
        print('Invalid choice. Please try again.')
```
